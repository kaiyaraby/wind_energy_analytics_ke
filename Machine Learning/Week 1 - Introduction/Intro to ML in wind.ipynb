## What is Machine Learning?
Machine Learning is an umbrella term used to describe models which 'learn' patterns or behaviours from training data and infer or estimate behaviours about new unseen data. In recent years machine learning has become integral to almost all analytics as data sets expand and access to computational power increases. Many machine learning concepts and models are surprisingly simple, but with a lot of the focus on complex generative AI topics, and a lot of confusing tech lingo it can be an intimidating field. This course will aim to cover the techniques from a beginner-friendly perspective. If you took and understood the stats part of the course, the hard part is over!

### Related Terms
In the field of Machine Learning many different terms are used to describe categories of techniques. A lot of these are very similar and models or sub-areas often fall into several categories.
- **Deep Learning**: more complex, multilayered models generally with the ability to capture more complex dynamics, patterns and behaviours. Most commonly, this term refers to more intricate architectures of Neural Networks (NNs) with more hidden layers.
- **Artificial Intelligence (AI)**: the ability of computers to carry out tasks in a way that mimics human intelligence. Colloquially, we often use AI now to refer to the most cutting edge or innovative models at the time. Models that were previously described as AI are often recategorised once they become more commonplace. Similarly, models that are considered to be 'black-box' or uninterpretable are often referred to AI models by default. When we talk about AI models in wind energy most commonly these will be some type of Neural Network model.
- **Data-driven models (DDMs)**: data-driven models are algorithms trained on historical data. This term is usually used to differentiate from physics-based, or physical, models with the implication that the model is largely based on information we have gleaned from the data rather than assumptions based on our understanding of the underlying system dynamics. These are often 'black-box' models which are less interpretable.
- **Physical or Physics-based models**: models defined by the true behaviour or dynamics of a system. They may be trained on data, but are designed logically in order to simulate physical phenoma that we understand. Because of this they are highly interpretable.
- **Surrogate models**: models which simulate or replace a more complex set of system dynamics, mapping inputs to outputs through a more simplistic model in a way that mimics the true relationship. Surrogate models can be physical or data-driven and often fall under the umbrella of ML or AI.
- **Statistical models**: these lie between data-driven and physical models. We generally make assumptions about the distributions or behaviours of certain variables, and use data to adapt these assumptions in order to build our final model. They are informed both by our understanding of the system and what we observe.

## Problem Types: Supervised, semi-supervised and unsupervised learning
Determining what type of problem we are working informs which types of model we select and use. Typically a problem is considered to be a supervised, semi-supervised or unsupervised learning problem. This is usually determined by what type of data we are working with.
#### Supervised Learning
Supervised learning involves working with *labelled* data sets. This means we have known output labels within our training data set. These types of data sets allow us to learn from set examples where all the information is available in order to accurately classify or predict outputs. We are also able to easily quantify measures such as accuracy and precision due to these available labels. The two main types of problems within supervised learning are:
- **Classification**: A classification problem is one where we wish to group data into different categories, or *classes*. Using labelled data our models learn what the inputs of points in each category look like and use this to infer which category unseen data points will fall into based on their inputs. Input data for these problems is *categorical* where there is a finite number of groups that can be assigned. One example in wind energy is classifying types of damage in components based on sensor readings or imagery.
- **Regression**: A regression problem generally involves *continuous* data, mapping numerical inputs to numerical outputs. The most simple type of regression is a linear regression with a single-dimensioned input and output, but we may have very complex regression methods with higher-dimensional models, inputs and outputs. Examples of this in wind energy include mapping wind conditions or sensor readings to expected remaining useful life, predicting weather conditions and high wind speeds, and inferring conditions at one turbine from another.

#### Unsupervised Learning
Unsupervised learning involves working with *unlabelled* data sets. This type of modelling involves a mixture of exploratory data analysis and inference modelling which can analyse relationships and similarities between data of unknown groups and use this to define clusters, sets or boundaries. Typically unsupervised learning problems are **Clustering** problems, where we wish to form groups within a population. For example, within wind we may wish to cluster together similar turbines on a site to simplify analysis, or to optimise where we place monitoring devices. Other unsupervised learning problems include **Association** which involves identifying similar data points, and **Dimensionality reduction** where we map high-dimensioned data into lower dimensions whilst maintaining the similarities and differences between points as best as possible.

#### Semi-supervised Learning
Semi-supervised learning generally involves having large sets of unlabelled data, where we may be able to add labels to some cases. For example, we may have many reported failures of a certain type, but only a handful of reports describing the sub-type of fault we wish to classify the data into. We can add a sub-type label to these data points, but many still have an unknown sub-type. This allows some learning to be carried out through examples but still relies on the ability to infer patterns from the remainder of the set.

## Machine Learning in Wind Energy
Machine learning has become increasingly popular in wind energy over the past 7 years, accelerating even further with the recent AI boom. Much of current research, acadaemia and industry innovation has focussed on developing and using ML and AI methods to improve data leveraging within the sector. With many streams of traditional, high-frequency numerical data such as weather and environmental data, high-resolution hardware monitoring & IoT (internet of things) setups and energy market price points, many traditional models can be transferred in a straightforward way from other sectors with high ML use such as trading. However, our industry is unique in the huge variety of different data sources that inform our work. As well as these traditional data sources, we work with complex geospatial data, imagery from site (satellite and engineering reports) and qualitative and natural-language information (e.g. site reports). As well as this, our data is often messy or incomplete (which we sometimes refer to as censored). This means that a lot of current wind energy machine learning work involves taking techniques from other fields and adapting them to our unique industry and problems. For example, we can borrow a lot of techniques from meteorological research to inform how we best make inferences and predictions from geospatial data, we can borrow from medical statistics and survival analysis to inform how we work with failure data, and we can borrow from computational linguistics to gain key information from qualitative reports, but all of these can be further researched and improved upon to tailor them to our domain.

### Common Model Types in Wind Energy
The most commonly used machine learning models in wind energy are Neural Networks (NNs), and this is what will take up the majority of our learning time during this course. Neural Networks are models built to mimic the human brain by sending information through interconnected layers of 'neurons'. These models are very flexible as the modeller can design the architecture and parameters in order to best approach different problems. Complex forms of neural networks known as transformer models are responsible for training many of the generative AI and large language model products on the market.
Other common methods within wind energy include Linear & Polynomial Regression, Logistic Regression, Decision Trees, Genetic Algorithms, Gaussian Processes, Support Vector Machines, Polynomial Chaos Expansion, Gradient Boosting, Dynamic Mode Decomposition, Bayesian Networks, Radial Basis Function Networks, and Random Forests. This course aims to give a brief introduction to these methods, and to explain how and why we use them.

### Common Use Cases in Wind Energy
As ML and AI are growing in use across the sector we are seeing many emerging use cases and discovering new ways to leverage data and modelling to improve how we manage wind assets so the possibilities are endless, but some of the most common examples of ML use in wind energy include:
- Weather forecasting
- Site selection
- Site layout optimisation
- Load estimation
- Remaining useful life estimation
- Early fault detection
- Maintenance strategy optimisation
- Wake modelling
- Wake steering
- Fault classification
- Root cause analysis
- Curtailment strategy optimisation
- Energy market forecasting and optimisation
- Risk and uncertainty quantification
- Grid management
