## 1 - Why we use packages
One of the most significant benefits of Python over other programming language is its broad range of open-source packages. Although Python comes well equipped with a lot of in-built functionality and building blocks to create useful code, packages provide us with ready-made tools that save us a lot of time. Packages are essentially collections of prewritten code that make Python much more useable. Instead of writing all functions from scratch to import, handle and visualise data, we can use packages such as pandas, NumPy and Matplotlib respectively, which are well-tested and standardised with helpful documentation available. By using packages such as these, we can vastly reduce the amount of code we need to write, speed up our processes, and make our code more reusable.

## 2 - Importing and using pre-installed packages
There are several ways to import a package for use. We are going to begin with the example package NumPy. This package is incredibly useful for carrying out mathematical functions in Python, and contains useful data structures that we will use regularly in this course. These functionalities will be covered in Section 3. To access the functionalities, we need to import the module. The most simple way to do this is to type import and then the name of the package as below.

```python
import numpy
```

We can then access and use methods and attributes of this package by prefacing them with the name of the package followed by a full stop. For example the code below runs the method sqrt() which takes the square root of a given number.

```python
numpy.sqrt(4)
```

Sometimes, we may not wish to type the full name of the package every time. Typically, we abbreviate it to 2 or 3 letters, in the case of numpy we generally use np. To import a package under a specific name we type import package as name, as below.

```python
import numpy as np
```

We may now access and use methods and attributes through its abbreviated name.

```python
np.sqrt(9)
```

In some cases, where we are only interested in one method of a package we can import that specifically using the syntax from package import method.

```python
from numpy import sqrt
```

Now we can simply use the name of the method without specifying the package.

```python
sqrt(16)
```

We can import multiple features by typing them out in a comma separated list.

```python
from numpy import sqrt, round
```

```python
round(0.23454, 2)
```

This method can also be used to import *all* methods from a package in this way by replacing method with * to represent all.

```python
from numpy import *
```

Not only has sqrt been imported, but additional methods and attributes. Below we can see the example attribute numpy.e, Euler's constant.

```python
e
```

We can also combine the syntax of 'from' and 'as' and type from package import method as name to import a method from a package to have a specific name to later reference.

```python
from numpy import sqrt as SQRT
```

```python
SQRT(25)
```

However, there is a big drawback with importing individual methods, and with naming them. Without the preface of the package this name may well have the same name as a built-in Python method. This can then overwrite that method and block functionality. The table below summarises these import methods and their advantages and disadvantages. Following this section we will study some example packages, and recommend the appropriate method for importing.


|Method|Syntax|Pros|Cons|
|---------|-----------|------|-----|
|1|import package|Simplest method|Can be inefficient to install all methods, retyping long name can reduce readability/reusability|
|2|import package as name|Packages that will be used often, and have multiple methods that will be used|Can be inefficient to install all methods|
|3|from package import method|Very efficient when only one or a few methods/attributes are required|Name may be not be ideal for reusability/readability|
|4|from package import method as name|Very efficient when only one or a few methods/attributes are required, short name more reusable/understandable|Must be careful not to overwrite existing function with name|
|5|from package import *|Least code to write|Inefficient to install all elements, names may overwrite in-built methods|

##### Adding, removing and sorting

An array may be sorted in numerical order using the sort() method. This is both an implicit method of an array which can be used to sort the array without returning anything, as below:

```python
x = np.array([2,1,2,3,4,5,2,1])
x.sort()
x
```

or by using the in-built NumPy method which returns the sorted array.

```python
np.sort(x)
```

View more sorting methods [here](https://numpy.org/devdocs/reference/routines.sort.html).

Once sorted, an array may be flipped to produce reverse-ordered data.

```python
np.flip(x)
```

Arrays can be combined using the concatenate() method, as long as their dimensions are suitable. The arrays to be combined must be given in a list.

```python
a = np.array([1,2,3])
b = np.array([4,5,6])
c = np.array([7,8])

np.concatenate([a,b,c])
```

```python
a = np.array([[1,2], [3,4], [5,6]])
b = np.array([[7,8], [9,10]])

np.concatenate([a,b])
```
