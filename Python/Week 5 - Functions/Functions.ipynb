A function is a block of code which only runs when it is called, we often use them when we wish to define a useful actions that we would like to carry out more than once. Rather than rewriting code every time, we can define it once in a function and call the function every time we would like to run it.

Python already has a number of in-built functions - some of these we have already seen in previous sessions for example *len()*, which tells us the number of items in a list

```python
len([1,2,3])
```

and *type()* which tells us the type of object

```python
type({1:'a'})
```

To use these functions we have to *call* them. We do this by typing the function name followed by parentheses containing the inputs to the function known as the *arguments*.

## 1: Defining a function
We define a function using the keyword *def* followed by the function name, parentheses which may contain optional inputs to the function, a colon, and then the code to enact indented by a tab (or three spaces). The example below defines a function which when called prints a sentence.

```python
def my_function():
    print('This function prints a sentence!')
```

Let's break down the components of the above code used to define the function.

|Number|Component|Description|
|---|---|---|
|1|def|The keyword that initialises the definition of the function|
|2|function name|A valid identifier$^1$ (name) that can be used to call the function|
|3|arguments|Parentheses which may either be empty, or contain the parameters and default values of arguments$^2$|
|4|Colon|Used to denote the end of the function header and initialise the start of the code block|
|5|Code block|Code indented by a tab (or three spaces) to be run every time the function is called|


[1] A *valid* identifier is a continuous string starting with a letter, and made up only of letters, numbers and the underscore character.

[2] Parameters are the identifiers/names used to define the function and how the arguments will be used, whilst the arguments are the user-input values that are used when the function is called

We can now *call* this function by typing the name of the function followed by parentheses.

```python
my_function()
```

# 2: Arguments
What if we wanted a function that printed the results of any two numbers multiplied together instead of just 2 and 16? We can do this by specifying *parameters* in the parentheses after the function name.

```python
def multiply(number_1, number_2):
    result = number_1*number_2
    print(result)
```

```python
number_1 = 3
number_2 = 9
multiply(number_1, number_2)
```

We see that when specifying the parameters we have used identifiers (names) so that we can use them in the function. Unlike the *function name* these do not have to be unique and can already be names of existing variables, within the function this identifier will be temporarily overwritten by the argument given to the function.

```python
number_1 = 2000
multiply(5,10)
multiply(35,7)
multiply(25,16)
```

Outside of the function the original value associated with the identifier is unchanged.

```python
print(number_1)
```

When we called the function we entered what we call *positional arguments*, this means specified them by typing them in the order they appear, without using their identifiers. When we open the brackets the first number entered will be interpreted as *number_1* and the second number entered will be interpreted as *number_2*. For this function it is fairly arbitrary which number comes first or second, but for many functions this is not the case.

```python
def exponentiate(number_1, number_2):
    result = number_1**number_2
    print(result)
```

This function raises the first number to the power of the second number, reordering the arguments now gives very different results.

```python
exponentiate(2,10)
exponentiate(10,2)
```

We can vary the order of the arguments whilst keeping the association the same by specifying the identifier using *parameter=argument*. Entering arguments in this way is known as using *keyword arguments*. Using keyword arguments removes the need for correct order and aids the readability of code for other users.

```python
exponentiate(number_1=2, number_2=10)
exponentiate(number_2=10, number_1=2)
```

## Default (optional) Arguments
In some cases we may want to have the flexibility of whether or not we specify an argument, by setting the argument to a default value. We do this by specifying parameter=default. The function below prints a string *n* times. By default (if no arguments are specified) 'Hello' is printed once.

```python
def printer(n=1, string='Hello'):
    for i in range(n):
        print(string)

printer()
```

We may specify neither (as above), one, or both of the arguments.

```python
printer(n=5)
```

```python
printer(string='A slightly longer sentence')
```

```python
printer(n=5, string='A slightly longer sentence')
```

Default arguments can be used alongside positional arguments - where some *must* be specified, and others are optional. When we define some as optional and others required, we **must** put all required parameters before any default when defining the function.

```python
def multiplier(number_1, number_2=10):
    result = number_1*number_2
    print(result)

multiplier(number_1=5)
```

Here we may either run the function specifying the first number, or we can specify both.

```python
multiplier(number_1=5, number_2=15)
```

Default arguments can also be used to allow a user to control an aspect of the code via a boolean operator. This could be to choose between formatting, or to determine whether a secondary function, or particular block of code is run.

```python
def multiplier(number_1, number_2, print_number_only=True):
    result = number_1*number_2
    if print_number_only:
        print(result)
    else:
        print(f'{number_1} multiplied by {number_2} is equal to {result}')

multiplier(number_1=3,number_2=7)
multiplier(number_1=3,number_2=7, print_number_only=False)
```

## Variable number of arguments

We may define a function with an optional number of arguments, where we are applying something to all arguments in the same way, by writing *args in the parentheses where we would usually specify parameters.

```python
def print_out(*args):
    print(*args)

print_out(1,2,3,4,5)
```

```python
def multiply_all(*args):
    total = 1
    for arg in args:
        total *= arg
    print(total)

multiply_all(1,2,3,4,5)
```

# 3: Return Values

The functions we have seen so far have all printed values. This is a type of *environment-affecting* function. These are functions which when called are reflected in the environment, either through having something printed, or otherwise impacting the way the environment or behaviour of the code outside the function. In general, functions can either affect the environment, return a value, or both. To return a value within a function we type return followed by the value we wish to return.

```python
def multiplier(number_1, number_2):
    result = number_1*number_2
    return result

multiplier(6,8)
```

Although the output might look very similar it has been *returned* rather than printed. This means we may assign it to a variable to be used later in the code.

```python
result = multiplier(6,8)
print(result)
print(result*10)
```

# 4: Documentation
Documenting code is essential to make sure that we ourselves and others can understand the code we have written. Week 13 will give a more comprehensive overview of why and how we can document our code correctly, but we will provide a brief overview here into best-practice for documentation of functions. There are two key ways we document written functions:
- Comments
- Docstrings

Comments are explanatory sentences starting with a #. The # allows following text to be recognised as a comment rather than as code, and are ignored by Python when the programme runs. The main purpose of comments is to give quick notes and explanations. These can be used to explain what a certain line, or loop, of code is doing, or why. They can be written either as a separate line (as below)

```python
def calculate_area(radius):
    # Calculates area of circle using the formula πr²
    return 3.14159 * radius * radius
```

or at the end of a line of code,

```python
def calculate_area(radius):
    return 3.14159 * radius * radius # Calculates area of circle using the formula πr²
```

but must not span more than a line.

A *docstring* (documentation string) is a string literal that appears as the first statement inside a function. Unlike comments, docstrings are stored as an attribute of the function which can be accessed to help us understand how the function works. They are meant to provide high-level documentation about what the function does, its parameters, return values, and usage. Docstrings are written using triple quotes (""" or ''') and can span multiple lines.

```python
def calculate_area(radius):
    """
    Calculates the area of a circle given its radius using the formula πr²

    Parameters:
        radius (float): The radius of the circle

    Returns:
        float: The calculated area of the circle
    """
    return 3.14159 * radius * radius
```

We may access the docstring by typing the name of the function followed by .__doc__

```python
print(calculate_area.__doc__)
```

In general for a function of this simplicity a single comment is likely to be sufficient, however for more complex functions a doc-string of this nature specifying functionality, parameters and return values with their type (e.g. float, int) and meaning, is considered good-practice.

Whilst both are used for documentation, comments are to explain the 'how' or 'why' of short chunks of code and are ignored by Python, whilst Docstrings explain what the function does and how to use it, and are accessible attributes.
