### 1.2: If statements
An *if* statement evaluates a statement, and if that statement is True it executes the code indented below. The syntax is the word 'if' followed by the statement to evaluate and a colon, then code below indented by three spaces (or a tab).


```python
if True:
    print('Statement is True')
```

```python
if False:
    print('Statement is True')
```

In the first scenario we see the code executes and 'Statement is True' is printed. This is because True is equivalent to True. However, if we change the statement to be False (not True), then the code is not executed and nothing is printed.

```python
true_statement = (2==2)
false_statement = (5<3)
```

```python
if true_statement and false_statement:
    print('Both statements are True')

if true_statement or false_statement:
    print('At least one of the statements is True')

if not true_statement and not false_statement:
    print('Both statements are False')

if not true_statement or not false_statement:
    print('At least one of the statements is False')
```

### Elif statements

An *elif* statement is a statement operator that evaluates a new statement if the if statement is False, executing new code only if the new statement is True.
It is short for 'else if'. 

```python
if false_statement:
    print('The first statement is True!')
elif true_statement:
    print('The first statement was False, but this new_statement is True!')
```

The *if* statement was not True, so the code was not executed and the first text was not printed. The *elif* statement was True, so the code in this section was executed and the second text was printed.

### Else Statements
An *else* statement is a statement which captures all other cases, executing code if the if (and elif statements) are all False.

```python
if false_statement:
    print('The first statement is True!')
else:
    print('The first statement is False!')
```

```python
second_false_statement = (7%2==2)

if false_statement:
    print('The first statement is True!')
elif second_false_statement:
    print('The second statement is True!')
else:
    print('None of the statements are True!')
```

An else statement *must* go at the end, after all other elif statements. You can have more than one elif statement.

## 2: For loops
A *for loop* is used for iterating over a number of different objects, meaning we can execute some code for each item in a list, set, tuple, dictionary, string, or range of values. The code below prints the ages in a list. We see that the syntax of the for loop allows us to define a symbol to refer to each item in the list as we iterate over it. In each instance, we use x to refer to the current item in the list, when we move on to the next item x then refers to this new item and 'forgets' the previous instance.

```python
age_list = [15, 39, 63]

for x in age_list:
    print(x)
```

This symbol can be a letter such as x or i, or can be a longer sequence like a word or phrase, which can be helpful for keeping track of the meaning of variables used.

```python
for age in age_list:
    print(age)
```

The memorylessness of this property means we cannot continuously add using this alone. For this reason, an additional variable can be defined outside of the for loop to keep track of certain traits of the set. For example, the code below adds the current age to the total age every iteration to allow us to add up all ages in the list.

```python
total_age = 0

for age in age_list:
    total_age += age

print(f'Total age = {total_age}')
```

#### Integrating if statements
We can use the *if* statements we previously learned about to alter variables only in certain cases. For example, we can work out the number of ages in the list exceeding 18.

```python
number_of_adults = 0

for age in age_list:
    if age >= 18:
        number_of_adults += 1

print(f'Number of adults = {number_of_adults}')
```

#### Breaks
Sometimes we may be iterating over very long lists, but only care about one parameter. For example, if we are interested in knowing whether all participants are over the age of 18, as soon as we find one age that is under this value we no longer need to iterate over the rest of the list. In this case we can use a *break* statement which exits the loop.

```python
iterations = 0
for age in age_list:
    iterations += 1
    if age < 18:
        print('A participant is under the age of 18')
        break

print(f'Number of iterations: {iterations}')
```

We see that this loop broke after the first iteration, this can really help efficiency when we have very long sequences to run through.

#### Types of Sequences
So far we have seen the list example but we can iterate over any of the types of sequences we learned in Week 1:
* Strings

```python
for letter in 'bananas':
    print(letter)
```

* Tuples

```python
for number in (1,2,3,4,5):
    print(number**2)
```

If we wish to run through a list or tuple of numbers such as above we can use the range() function to define a range of integer numbers. The simplest way to use the range function is to specify the end number. This results in iterating over every integer number from 0 until one less than the end number. I.e. range(5) will iterate through 0,1,2,3,4 but not reach 5.

```python
for i in range(5):
    print(i**2)
```

If we don't want to start from 0, we can specify both a start and end number.

```python
for i in range(2,5):
    print(i)
```

And we can even specify a step change other than 1, for example to cover all even numbers.

```python
for i in range(2,11,2):
    print(i)
```

* Dictionaries

The code will assume we are iterating over the *keys* of the dictionary

```python
dictionary = {1:'a', 2:'b', 3:'c', 4:'d', 5:'e'}

for key in dictionary:
    print(key, dictionary[key])
```

but we can iterate specifically over the values of the dictionary by typing dictionary.values()

```python
for value in dictionary.values():
    print(value)
```

* Sets

```python
for item in {1,2,3,'a'}:
    print(item)
```

#### Nested for loops
In some cases we wish to use multiple for loops, for example to work with subparts or smaller parts of objects within the sequence.

```python
sentence = 'It is a nice day today'

words_with_is = 0

for word in sentence:
    for letter in word:
        if letter == 'i' or letter == 'I':
            words_with_is += 1

print(words_with_is)
```

When we use break statements within a nested (sub) for loop, we only break out of that part.

```python
for fruit in ['apple', 'banana', 'tangerine']:
    for number in [1,2,3,4,5]:
        print(fruit, number)
        if number >= 3:
            break
```

We see that the number loop is always broken at 3, but we still continue to iterate over the remaining fruits after this point breaking at each point in that iteration where the number is exceeded.

## 3: While loops
A *while loop* is used to continuously iterate over code while a certain condition is being met. We can use this to run equivalent loops to those in range by first defining a variable and adding one to the variable every time until we exceed a value of interest.

```python
i = 0

while i<6:
    print(i)
    i += 1
```

We can also run while True to run continuously. A break statement is essential if we run code in this format, or the code will run forever!

```python
i = 0
while True:
    print(i)
    i += 1
    if i==6:
        break
```

Continue statements also rely heavily on a second useful statement known as a continue statement. This ends the current iteration, and moves on to the next. For example the code below skips printing the even numbers by moving on to the next iteration before printing if i is divisible by 2.

```python
i = 0
while i < 10:
    i += 1
    if i%2 == 0:
        continue
    print(i)
```

One of the most relevant uses of while loops in engineering is for optimization, to run mathematical functions until we find an answer which meets certain conditions or falls within a certain closeness to a True value. We will be able to see this more in practice later in the course. The table below (based on [GeeksforGeeks](https://www.geeksforgeeks.org/python/difference-between-for-loop-and-while-loop-in-python/)) summarises the key differences and use cases between for loops and while loops.

||For loop|While loop|
|---|---|---|
|Use Case|A For loop is designed to iterate over a sequence such as a list, tuple, dictionary, string or range.|A While loop is used when the number of iterations is not known in advance or when we want to repeat a block of code until a certain condition is met. Generally suitable for more complex situations such as optimisation.|
|Requirements|A For loop requires a sequence to iterate over|A While loop requires an initial statement to evaluate the Truth value of|
|Efficiency|A For loop is more efficient than a while loop when iterating over sequences, since the number of iterations is predetermined and the loop can be optimized accordingly|A While loop may be more efficient in certain situations where the condition being tested can be evaluated quickly|
