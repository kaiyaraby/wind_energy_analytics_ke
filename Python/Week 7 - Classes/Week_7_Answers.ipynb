##  What are classes?
Classes are fundamental to *object oriented programming* (OOP), a programming model that organizes software design around objects, which contain both data (*attributes*) and methods (*actions*) that operate on that data, enabling modular, reusable, and maintainable code. Classes can be thought of as blueprints for making lots of things that are identical. Let's imagine we have made a class called 'Room', which defines a blueprint for any room. Once we have the blueprint we can create objects, known as *instances* by calling the class like this:

kitchen = Room()

hall = Room()

bedroom = Room()

Once defined, we can find out information about these instances due to their inherent *attributes*. These are accessed as 'object.attribute' for example some key information we might like to store about a room might be 

when working with a class we can define an attribute by referring to 'self' which will be the instance of the class. For example, to set the kitchen height we would type within the class 'self.height = *** '. Some of this information will need to be entered by the user. Unlike functions where we define in terms of the information put in as 

```python
def function(a,b,c):
 return None
```

we simply define class by it's name

```python
class Room:
    pass
```

and define methods within it that take information. A *method* is like a function, but it can only be applied to the object itself. As such whenever we define a method within a class the key argument is 'self', following that we may add additional arguments which will be input by the user when called. We'll start by discussing the most important method, the \_\_init\_\_ method. This method is called automatically when an instance is created, and takes in the key information. For example, let's write a very simple class for Room, taking in only the key measurements which will apply to every room: width, height and length.

```python
class Room:
    def __init__(self, width, height, length):
        self.width = width
        self.height = height
        self.length = length
```

Now when we create our room object we have defined some key attributes about it, and can access them as discussed before.

```python
kitchen = Room(width=20, height=10, length=30)

width = kitchen.width

print(f'Kitchen Width = {width} m')
```

We may want to combine these measurements to work out the area of the room, we can do this by defining a method called calc_area. Since the object already contains all the information we need, all we need as an argument is self.

```python
class Room:
    def __init__(self, width, height, length):
        self.width = width
        self.height = height
        self.length = length

    def calc_area(self):
        return self.width*self.height*self.length
```

We can then access the method by using object.method()

```python
kitchen = Room(width=20, height=10, length=30)
area = kitchen.calc_area()
print(f'Area: {area} m^3')
```

For functions like this where we do not need to input any information ourselves, we can use the decorator @property. This allows us to define it such that it can be called as an attribute.

```python
class Room:
    def __init__(self, width, height, length):
        self.width = width
        self.height = height
        self.length = length
    @property
    def calc_area(self):
        return self.width*self.height*self.length
```

```python
kitchen = Room(width=20, height=10, length=30)
area = kitchen.calc_area
print(f'Area: {area} m^3')
```

Finally we may wish to calculate this as soon as we call the function. We can do this by running the function within the \_\_init\_\_ statement so that it is called as soon as the object is initialized.

```python
class Room:
    def __init__(self, width, height, length):
        self.width = width
        self.height = height
        self.length = length
        self.area = self.calc_area
        
    @property
    def calc_area(self):
        return self.width*self.height*self.length
```

```python
kitchen = Room(width=20, height=10, length=30)
area = kitchen.area
print(f'Area: {area} m^3')
```

We can also define class methods which require more information entered by the user.

```python
class Room:
    def __init__(self, width, height, length):
        self.width = width
        self.height = height
        self.length = length
        self.area = self.calc_area
        
    @property
    def calc_area(self):
        return self.width*self.height*self.length

    def calc_area_per_person(self, n_people):
        return self.calc_area/n_people
```

# Inheritance
We can define more specific blueprints by defining a child class which inherits all of its properties and methods from its parent class as follows

for example let's create a child class of our Room class for bedrooms

```python
class BedRoom(Room):
    pass
```

We can then set up the class to initialize using a call to its parent function.

```python
class BedRoom(Room):
    def __init__(self, width, height, length):
        Room.__init__(self, width, height, length)
```

This can also be done using the super() method. This method does *not* take self as an argument

```python
class BedRoom(Room):
    def __init__(self, width, height, length):
        super().__init__(width, height, length)
```

But we might want to add more required information for this child class, for example number of beds.

```python
class BedRoom(Room):
    def __init__(self, width, height, length, n_beds):
        super().__init__(width, height, length)
        self.n_beds = 2
```

```python
bedroom_1 = BedRoom(width=20, height=15, length=20, n_beds=2)
```

bedroom_1 now has the same information and methods available as its parent class

```python
area = bedroom_1.calc_area
print(f'Bedroom area = {area} m^3')
```

but also additional functionality specified just to that class

```python
beds = bedroom_1.n_beds
print(f'Number of beds = {beds}')
```

This allows us to build more and more specific child classes without needing to redefine the basic mould. One important note is that this child class takes precedent over anything previousely set. Using methods or attributes with the same name as those defined in the parent class will be overriden. Let's have a look at an example of a full script.

```python
class Room:
    def __init__(self, width, height, length):
        self.width = width
        self.height = height
        self.length = length
        self.area = self.calc_area

    @property
    def calc_area(self):
        return self.width*self.length*self.height

    @property
    def calc_floorspace(self):
        return self.width*self.height

    def calc_floorspace_per_person(self, n_people):
        floorspace = self.calc_floorspace
        return floorspace/n_people

class BedRoom(Room):
    def __init__(self, width, height, length, n_beds):
        super().__init__(width, height, length)
        self.n_beds = n_beds

class LivingRoom(Room):
    def __init__(self, width, height, length, n_seats):
        super().__init__(width, height, length)
        self.n_seats = n_seats


kitchen = Room(10, 20, 10)
bedroom_1 = BedRoom(20, 20, 30, 2)
bedroom_2 = BedRoom(15, 15, 40, 1)
living_room = LivingRoom(30, 15, 20, 2)

kitchen_area = kitchen.area
print(f'Kitchen area: {kitchen_area} m^3')

kitchen_floorspace = kitchen.calc_floorspace
print(f'Kitchen floorspace per person: {kitchen_floorspace} m^2')

kitchen_floorspace_pp = kitchen.calc_floorspace_per_person(5)
print(f'Kitchen floorspace per person: {kitchen_floorspace_pp} m^2')
```

#### Extension - Documentation
Similarly to functions, it is helpful to document our classes to ensure they are understandable by other users. We will learn how to document classes in full detail in Week 13 but a general rule is that an overall class should be documented with a docstring, with methods documented using comments only.

Use the guidance in the documentation part of Week 5's tutorial to comment your class written in Exercise 5.

```python
class Vehicle:
    """
    A base class representing a generic vehicle.

    Attributes:
        n_seats (int): Number of seats in the vehicle.
        n_wheels (int): Number of wheels.
        max_speed (float): Maximum speed of the vehicle in mph.
    """

    def __init__(self, n_seats, n_wheels, max_speed):
        # Initialize a Vehicle object.
        self.n_seats = n_seats
        self.n_wheels = n_wheels
        self.max_speed = max_speed


class Car(Vehicle):
    """
    A class representing a car, inheriting from Vehicle.

    Attributes:
        mpg (float): Fuel efficiency in miles per gallon.
    """

    def __init__(self, n_seats, n_wheels, max_speed, mpg):
        # Initialize a Car object.
        super().__init__(n_seats, n_wheels, max_speed)
        self.mpg = mpg

    def calc_fastest_time(self, distance):
        # Calculates the fastest possible travel time for a given distance.
        return distance / self.max_speed


# Example usage
car = Car(n_seats=5, n_wheels=4, max_speed=100, mpg=30)
print(car.calc_fastest_time(200))  # Output: 2.0 (hours)
```
